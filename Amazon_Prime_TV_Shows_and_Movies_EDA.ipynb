## 3. ***Data Wrangling***

### Data Wrangling Strategy & Planned Manipulations

Before performing exploratory visual analysis, the datasets must undergo rigorous cleaning, transformation, and structural wrangling:

1. **Deduplication:**
   - Remove 3 duplicate rows from `titles_df` and 56 duplicate rows from `credits_df` to ensure total data integrity.
2. **String List Evaluation & Unnesting:**
   - Parse `genres` and `production_countries` columns from string representations into true Python lists using `ast.literal_eval`.
   - Replace empty string lists (`'[]'`) with explicit `'Unspecified'` values to avoid missing counts during unnesting.
3. **Handling Missing & Structural Values:**
   - Standardize `seasons` for `MOVIE` content as `0` or `Not Applicable` while preserving integer values for `SHOW` titles.
   - Impute missing text descriptions and fill non-rated `age_certification` entries with `'Unrated'`.
4. **Relational Dataset Integration:**
   - Merge `titles_df` and `credits_df` on `id` to create a consolidated dataset (`merged_df`) enabling multi-dimensional talent analytics (top actors, top directors, ratings by director).

### Data Wrangling Code

```python
# 1. Deduplication
titles_clean = titles_df.drop_duplicates().copy()
credits_clean = credits_df.drop_duplicates().copy()

# Handle duplicate title IDs consistently by keeping the first occurrence
titles_clean = titles_clean.drop_duplicates(subset=['id'], keep='first').copy()

print(f"Cleaned Titles Shape: {titles_clean.shape}")
print(f"Cleaned Credits Shape: {credits_clean.shape}")

# 2. Parsing Stringified Python Lists for Genres and Production Countries
def parse_list_column(val):
    if pd.isna(val) or val == '[]' or val == '':
        return ['Unspecified']
    try:
        parsed = ast.literal_eval(val)
        return parsed if isinstance(parsed, list) and len(parsed) > 0 else ['Unspecified']
    except (ValueError, SyntaxError):
        return ['Unspecified']

titles_clean['genres_list'] = titles_clean['genres'].apply(parse_list_column)
titles_clean['countries_list'] = titles_clean['production_countries'].apply(parse_list_column)

# 3. Categorical Missing Value Imputation
# Preserve structural missingness in 'seasons' for MOVIE records (kept as NaN)
titles_clean['age_certification'] = titles_clean['age_certification'].fillna('Unrated')
titles_clean['description'] = titles_clean['description'].fillna('No description available.')

# 4. Merging Titles and Credits Datasets on 'id'
merged_df = pd.merge(titles_clean, credits_clean, on='id', how='left')

# Verification check to confirm cleaned DataFrames exist
assert 'titles_clean' in locals(), "titles_clean DataFrame missing!"
assert 'credits_clean' in locals(), "credits_clean DataFrame missing!"
assert 'merged_df' in locals(), "merged_df DataFrame missing!"

print("SUCCESS: Data Wrangling executed successfully!")
print(f"Cleaned Titles Shape: {titles_clean.shape[0]} rows, {titles_clean.shape[1]} columns")
print(f"Cleaned Credits Shape: {credits_clean.shape[0]} rows, {credits_clean.shape[1]} columns")
print(f"Merged Dataset Shape: {merged_df.shape[0]} rows, {merged_df.shape[1]} columns")
print("Verification: titles_clean, credits_clean, and merged_df exist and are ready for analysis.")
```

### What all manipulations have you done and insights you found?

To ensure data integrity, facilitate accurate grouping, and support granular visualization across all 15 charts, the following systematic data wrangling and transformation operations were executed:

1. **Deduplication:**
   - Evaluated full-row duplicate records using `.duplicated().sum()`.
   - Dropped 3 duplicate rows from `titles_df` (reducing title count from 9,871 to 9,868).
